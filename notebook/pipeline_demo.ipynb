```python
import pandas as pd
import numpy as np

# Create sample PCOS dataset for demonstration
np.random.seed(42)
n_samples = 100

df = pd.DataFrame({
    'Age_(years)': np.random.randint(20, 50, n_samples),
    'Weight_(Kg)': np.random.uniform(45, 100, n_samples),
    'Height(Cm)': np.random.uniform(150, 185, n_samples),
    'BMI': np.random.uniform(18, 35, n_samples),
    'Blood_Pressure': np.random.choice(['Normal', 'High', 'Low'], n_samples),
    'AMH_(ng/mL)': np.random.uniform(0.5, 10, n_samples),
    'FSH(mIU/mL)': np.random.uniform(2, 12, n_samples),
    'LH(mIU/mL)': np.random.uniform(3, 80, n_samples),
    'Testosterone(ng/ml)': np.random.uniform(20, 80, n_samples),
    'PCOS_(Y/N)': np.random.choice([0, 1], n_samples)
})

# Add BMI Category
df['BMI_Category'] = pd.cut(df['BMI'], bins=[0, 18.5, 25, 30, 100], 
                             labels=['Underweight', 'Normal', 'Overweight', 'Obese'])

df.head()
```

```python
df["BMI_Category"].value_counts().plot(kind="bar")
```

```python
# Select only numeric columns for correlation analysis
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols].corr()["PCOS_(Y/N)"].sort_values().plot(kind="barh")
```
